```python
import pandas as pd

# Path to your text file
filepath = r"C:\Users\rkroker\Documents\Code\Python\MS PE Grading Correction\PE MS Grades.txt"

# Specific course ID to filter
PE6_filter = "PE6"
PE7_filter = "PE7"
PE8_filter = "PE8"

# Read the data from the text file
df = pd.read_csv(filepath, delimiter="\t")

# Filter data for the specific course ID
PE6_filtered = df[df['STOREDGRADES.Course_Number'] == PE6_filter]
PE7_filtered = df[df['STOREDGRADES.Course_Number'] == PE7_filter]
PE8_filtered = df[df['STOREDGRADES.Course_Number'] == PE8_filter]

# Calculate the average percent for the filtered data
PE6_course_avg = PE6_filtered.groupby(['STOREDGRADES.StudentID', 'STOREDGRADES.Course_Number'])['STOREDGRADES.Percent'].mean()
PE7_course_avg = PE7_filtered.groupby(['STOREDGRADES.StudentID', 'STOREDGRADES.Course_Number'])['STOREDGRADES.Percent'].mean()
PE8_course_avg = PE8_filtered.groupby(['STOREDGRADES.StudentID', 'STOREDGRADES.Course_Number'])['STOREDGRADES.Percent'].mean()

# Path to the output CSV file (replace with your desired name and location)
PE6_file = "student_average_PE6.csv"
PE7_file = "student_average_PE7.csv"
PE8_file = "student_average_PE8.csv"

# Export the result to a new CSV file
PE6_course_avg.to_csv(PE6_file)
PE7_course_avg.to_csv(PE7_file)
PE8_course_avg.to_csv(PE8_file)

print(f"Average percent for course '{PE6_filtered}' per student saved to: {PE6_file}")
print(f"Average percent for course '{PE7_filtered}' per student saved to: {PE7_file}")
print(f"Average percent for course '{PE8_filtered}' per student saved to: {PE8_file}")
```
