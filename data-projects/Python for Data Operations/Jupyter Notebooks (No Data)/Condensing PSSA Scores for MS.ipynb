```python
import pandas as pd
import os

#Automatic Directory Retrieval
# Get the user's current working directory
dir = os.getcwd() # C:\Users\rkroker\Documents\Code\Python\Condensing PSSA Scores for MS

# Combine the current path and file name using os.path.join
# data_path = os.path.join(dir, "linkit_stored_grades.csv")
# Charles Boehm (255) Files
CB_MA_Scores_File = os.path.join(dir, r"Data\255\Student Test Scores - PSSA Math - 23-24.csv")
CB_MA_Sections_File = os.path.join(dir, r"Data\255\Students - Class Roster - MA - 23-24.csv")
CB_REA_Scores_File = os.path.join(dir, r"Data\255\Student Test Scores - PSSA Reading - 23-24.csv")
CB_REA_Sections_File = os.path.join(dir, r"Data\255\Students - Class Roster - ENG - 23-24.csv")
# Pennwood (260) Files
PW_MA_Scores_File = os.path.join(dir, r"Data\260\Student Test Scores - PSSA Math - 23-24.csv")
PW_MA_Sections_File = os.path.join(dir, r"Data\260\Students - Class Roster - MA - 23-24.csv")
PW_REA_Scores_File = os.path.join(dir, r"Data\260\Student Test Scores - PSSA Reading - 23-24.csv")
PW_REA_Sections_File = os.path.join(dir, r"Data\260\Students - Class Roster - ENG - 23-24.csv")
# William Penn (265) Files
WP_MA_Scores_File = os.path.join(dir, r"Data\265\Student Test Scores - PSSA Math - 23-24.csv")
WP_MA_Sections_File = os.path.join(dir, r"Data\265\Students - Class Roster - MA - 23-24.csv")
WP_REA_Scores_File = os.path.join(dir, r"Data\265\Student Test Scores - PSSA Reading - 23-24.csv")
WP_REA_Sections_File = os.path.join(dir, r"Data\265\Students - Class Roster - ENG - 23-24.csv")

# File paths to the CSV files
# MA_Scores_File = r"C:\Users\rkroker\Documents\Code\Python\Condensing PSSA Scores for MS\Data\255\Student Test Scores - PSSA Math - 23-24.csv"
# MA_Sections_File = r"C:\Users\rkroker\Documents\Code\Python\Condensing PSSA Scores for MS\Data\255\Students - Class Roster - MA - 23-24.csv"
# REA_Scores_File = r"C:\Users\rkroker\Documents\Code\Python\Condensing PSSA Scores for MS\Data\255\Student Test Scores - PSSA Reading - 23-24.csv"
# REA_Sections_File = r"C:\Users\rkroker\Documents\Code\Python\Condensing PSSA Scores for MS\Data\255\Students - Class Roster - ENG - 23-24.csv"

# Assign Dataframes to import files
# Charles Boehm (255) Files
cb_ma_scores = pd.read_csv(CB_MA_Scores_File)
cb_ma_sections = pd.read_csv(CB_MA_Sections_File)
cb_rea_scores = pd.read_csv(CB_REA_Scores_File)
cb_rea_sections = pd.read_csv(CB_REA_Sections_File)
# Pennwood (260) Files
pw_ma_scores = pd.read_csv(PW_MA_Scores_File)
pw_ma_sections = pd.read_csv(PW_MA_Sections_File)
pw_rea_scores = pd.read_csv(PW_REA_Scores_File)
pw_rea_sections = pd.read_csv(PW_REA_Sections_File)
# William Penn (265) Files
wp_ma_scores = pd.read_csv(WP_MA_Scores_File)
wp_ma_sections = pd.read_csv(WP_MA_Sections_File)
wp_rea_scores = pd.read_csv(WP_REA_Scores_File)
wp_rea_sections = pd.read_csv(WP_REA_Sections_File)
```

```python
# Create a list of columns to drop
scores_columns_to_drop = ["Term ID", "Test ID", "Score Name", "Pct Score", "Alpha Score"]
sections_columns_to_drop = ["SID", "School", "Grade", "Gender", "Track", "Ethnicity", "Exp.", "Term", "Room", "Teacher Number", "Student Enroll Date", "Student Left Date"]

# Drop the columns and assign the result to a new dataframe (recommended)
# Charles Boehm (255)
cb_ma_scores_df = cb_ma_scores.drop(scores_columns_to_drop, axis=1)
cb_rea_scores_df = cb_rea_scores.drop(scores_columns_to_drop, axis=1)
cb_ma_sections_df = cb_ma_sections.drop(sections_columns_to_drop, axis=1)
cb_rea_sections_df = cb_rea_sections.drop(sections_columns_to_drop, axis=1)
# Pennwood (260)
pw_ma_scores_df = pw_ma_scores.drop(scores_columns_to_drop, axis=1)
pw_rea_scores_df = pw_rea_scores.drop(scores_columns_to_drop, axis=1)
pw_ma_sections_df = pw_ma_sections.drop(sections_columns_to_drop, axis=1)
pw_rea_sections_df = pw_rea_sections.drop(sections_columns_to_drop, axis=1)
# William Penn (265)
wp_ma_scores_df = wp_ma_scores.drop(scores_columns_to_drop, axis=1)
wp_rea_scores_df = wp_rea_scores.drop(scores_columns_to_drop, axis=1)
wp_ma_sections_df = wp_ma_sections.drop(sections_columns_to_drop, axis=1)
wp_rea_sections_df = wp_rea_sections.drop(sections_columns_to_drop, axis=1)

# Filter to exclude rows with Num Score = 0, 1, 2, 3 or 4
# Charles Boehm (255)
cb_ma_scores_df = cb_ma_scores_df[~cb_ma_scores_df['Num Score'].isin([0, 1, 2, 3, 4])]
cb_rea_scores_df = cb_rea_scores_df[~cb_rea_scores_df['Num Score'].isin([0, 1, 2, 3, 4])]
# Pennwood (260)
pw_ma_scores_df = pw_ma_scores_df[~pw_ma_scores_df['Num Score'].isin([0, 1, 2, 3, 4])]
pw_rea_scores_df = pw_rea_scores_df[~pw_rea_scores_df['Num Score'].isin([0, 1, 2, 3, 4])]
# William Penn (265)
wp_ma_scores_df = wp_ma_scores_df[~wp_ma_scores_df['Num Score'].isin([0, 1, 2, 3, 4])]
wp_rea_scores_df = wp_rea_scores_df[~wp_rea_scores_df['Num Score'].isin([0, 1, 2, 3, 4])]
```

```python
# Using the Score Conversion chart in place by the PA Board of Education
# https://www.stateboard.education.pa.gov/Documents/About%20the%20Board/Board%20Actions/2015/PSSA%20Mathematics%20Performance%20Level%20Descriptors.pdf

def assign_math_proficiency(row):
  grade_level = row['Test Grade Level']
  num_score = row['Num Score']
  
  if grade_level == 6:
    if num_score < 896:
      return 1, "Below Basic"
    elif 897 <= num_score <= 999:
      return 2, "Basic"
    elif 1000 <= num_score <= 1104:
      return 3, "Proficient"
    else:
      return 4, "Advanced"
  elif grade_level == 7:
    if num_score < 903:
      return 1, "Below Basic"
    elif 904 <= num_score <= 999:
      return 2, "Basic"
    elif 1000 <= num_score <= 1108:
      return 3, "Proficient"
    else:
      return 4, "Advanced"
  elif grade_level == 8:
    if num_score < 905:
      return 1, "Below Basic"
    elif 906 <= num_score <= 999:
      return 2, "Basic"
    elif 1000 <= num_score <= 1107:
      return 3, "Proficient"
    else:
      return 4, "Advanced"
  else:
    # Handle cases where Test Grade Level is not 6, 7 or 8 (optional)
    return None, None  # Or set default values

def assign_reading_proficiency(row):
  grade_level = row['Test Grade Level']
  num_score = row['Num Score']
  
  if grade_level == 6:
    if num_score < 874:
      return 1, "Below Basic"
    elif 875 <= num_score <= 999:
      return 2, "Basic"
    elif 1000 <= num_score <= 1114:
      return 3, "Proficient"
    else:
      return 4, "Advanced"
  elif grade_level == 7:
    if num_score < 844:
      return 1, "Below Basic"
    elif 845 <= num_score <= 999:
      return 2, "Basic"
    elif 1000 <= num_score <= 1129:
      return 3, "Proficient"
    else:
      return 4, "Advanced"
  elif grade_level == 8:
    if num_score < 885:
      return 1, "Below Basic"
    elif 886 <= num_score <= 999:
      return 2, "Basic"
    elif 1000 <= num_score <= 1129:
      return 3, "Proficient"
    else:
      return 4, "Advanced"
  else:
    # Handle cases where Test Grade Level is not 6, 7 or 8 (optional)
    return None, None  # Or set default values

# Create new columns with direct assignment (avoiding unpacking)
# Charles Boehm (255)
cb_ma_scores_df['Prof Level'] = cb_ma_scores_df.apply(lambda row: assign_math_proficiency(row)[0], axis=1)
cb_ma_scores_df['Prof Alpha'] = cb_ma_scores_df.apply(lambda row: assign_math_proficiency(row)[1], axis=1)
cb_rea_scores_df['Prof Level'] = cb_rea_scores_df.apply(lambda row: assign_reading_proficiency(row)[0], axis=1)
cb_rea_scores_df['Prof Alpha'] = cb_rea_scores_df.apply(lambda row: assign_reading_proficiency(row)[1], axis=1)
# Pennwood (260)
pw_ma_scores_df['Prof Level'] = pw_ma_scores_df.apply(lambda row: assign_math_proficiency(row)[0], axis=1)
pw_ma_scores_df['Prof Alpha'] = pw_ma_scores_df.apply(lambda row: assign_math_proficiency(row)[1], axis=1)
pw_rea_scores_df['Prof Level'] = pw_rea_scores_df.apply(lambda row: assign_reading_proficiency(row)[0], axis=1)
pw_rea_scores_df['Prof Alpha'] = pw_rea_scores_df.apply(lambda row: assign_reading_proficiency(row)[1], axis=1)
# William Penn (265)
wp_ma_scores_df['Prof Level'] = wp_ma_scores_df.apply(lambda row: assign_math_proficiency(row)[0], axis=1)
wp_ma_scores_df['Prof Alpha'] = wp_ma_scores_df.apply(lambda row: assign_math_proficiency(row)[1], axis=1)
wp_rea_scores_df['Prof Level'] = wp_rea_scores_df.apply(lambda row: assign_reading_proficiency(row)[0], axis=1)
wp_rea_scores_df['Prof Alpha'] = wp_rea_scores_df.apply(lambda row: assign_reading_proficiency(row)[1], axis=1)
```

```python
# Merge the dataframes using the StudentNumber column from df_ma_scores and the LID column from df_ma_sections
# Charles Boehm (255)
cb_merged_ma_df = pd.merge(cb_ma_scores_df, cb_ma_sections_df, left_on='StudentNumber', right_on='LID')
cb_merged_rea_df = pd.merge(cb_rea_scores_df, cb_rea_sections_df, left_on='StudentNumber', right_on='LID')
# Pennwood (260)
pw_merged_ma_df = pd.merge(pw_ma_scores_df, pw_ma_sections_df, left_on='StudentNumber', right_on='LID')
pw_merged_rea_df = pd.merge(pw_rea_scores_df, pw_rea_sections_df, left_on='StudentNumber', right_on='LID')
# William Penn (265)
wp_merged_ma_df = pd.merge(wp_ma_scores_df, wp_ma_sections_df, left_on='StudentNumber', right_on='LID')
wp_merged_rea_df = pd.merge(wp_rea_scores_df, wp_rea_sections_df, left_on='StudentNumber', right_on='LID')

# Create a list of columns to drop
merged_columns_to_drop = ["Student", "LID"]

# Charles Boehm (255)
cb_merged_ma_df = cb_merged_ma_df.drop(merged_columns_to_drop, axis=1)
cb_merged_rea_df = cb_merged_rea_df.drop(merged_columns_to_drop, axis=1)
# Pennwood (260)
pw_merged_ma_df = pw_merged_ma_df.drop(merged_columns_to_drop, axis=1)
pw_merged_rea_df = pw_merged_rea_df.drop(merged_columns_to_drop, axis=1)
# William Penn (265)
wp_merged_ma_df = wp_merged_ma_df.drop(merged_columns_to_drop, axis=1)
wp_merged_rea_df = wp_merged_rea_df.drop(merged_columns_to_drop, axis=1)
```

```python
# Create an Excel writer object with the desired filename
cb_writer = pd.ExcelWriter('255 PSSA Scores and Teachers.xlsx', engine='xlsxwriter')
pw_writer = pd.ExcelWriter('260 PSSA Scores and Teachers.xlsx', engine='xlsxwriter')
wp_writer = pd.ExcelWriter('265 PSSA Scores and Teachers.xlsx', engine='xlsxwriter')

# Write dataframes to seperate tabs within the excel document
# Charles Boehm (255)
cb_merged_ma_df.to_excel(cb_writer, sheet_name='Math Scores', index=False)
cb_merged_rea_df.to_excel(cb_writer, sheet_name='Reading Scores', index=False)
# Pennwood (260)
pw_merged_ma_df.to_excel(pw_writer, sheet_name='Math Scores', index=False)
pw_merged_rea_df.to_excel(pw_writer, sheet_name='Reading Scores', index=False)
# William Penn (265)
wp_merged_ma_df.to_excel(wp_writer, sheet_name='Math Scores', index=False)
wp_merged_rea_df.to_excel(wp_writer, sheet_name='Reading Scores', index=False)


# Save the changes to the Excel files
cb_writer.close()
print("CB Dataframes exported to 255 PSSA Scores and Teachers.xlsx!")
pw_writer.close()
print("PW Dataframes exported to 260 PSSA Scores and Teachers.xlsx!")
wp_writer.close()
print("WP Dataframes exported to 265 PSSA Scores and Teachers.xlsx!")
```
