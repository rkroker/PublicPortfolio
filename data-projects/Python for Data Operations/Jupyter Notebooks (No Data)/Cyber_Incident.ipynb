```python
import pandas as pd

# Load the CSV file into a DataFrame

# Replace 'your_file.csv' with the path to your CSV file
students = pd.read_csv('Students.csv', low_memory=False)
teachers = pd.read_csv('Teachers.csv', low_memory=False)
```

```python
# List all the column names
print("Columns in the DataFrame:", students.columns.tolist())
```

```python
# List all the column names
print("Columns in the DataFrame:", teachers.columns.tolist())
```

```python
# Save ALL column names to CSV files
students.columns.to_series().to_csv("student_columns.csv", index=False, header=False)
teachers.columns.to_series().to_csv("teacher_columns.csv", index=False, header=False)

print("student columns saved to 'student_columns.csv' teacher columns saved to 'teacher_columns.csv'")
```

```python
# Filter columns with data populated (non-null)
student_non_null_columns = students.columns[students.notna().any()]
teacher_non_null_columns = teachers.columns[teachers.notna().any()]
```

```python
# Print and Review Student non-null columns
print("Columns with data:", student_non_null_columns.tolist())
```

```python
# Print and Review Teacher non-null columns
print("Columns with data:", teacher_non_null_columns.tolist())
```

```python
# Save the non-null column names to a CSV files
student_non_null_columns.to_series().to_csv("student_non_null_columns.csv", index=False, header=False)
teacher_non_null_columns.to_series().to_csv("teacher_non_null_columns.csv", index=False, header=False)

print("student_non-null columns saved to 'student_non_null_columns.csv' teacher_non-null columns saved to 'teacher_non_null_columns.csv'")
```
