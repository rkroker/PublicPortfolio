```python
import pandas as pd

# List of columns to drop
columns_to_drop = [
    "SID", "School", "Gender", "Track", "Ethnicity", 
    "Course Dept", "Course", "Course Number", "Section", 
    "Exp.", "Term", "Room", "Teacher Number", 
    "Student Enroll Date", "Student Left Date"
]

merge_prep = [
    "Student", "Grade", "SID", "School", "Gender", "Track", "Ethnicity", 
    "Course Dept", "Course", "Course Number", "Section", 
    "Exp.", "Term", "Room", "Teacher Number", 
    "Student Enroll Date", "Student Left Date"
]

# Load the four CSV files and drop unessacary columns
ENG = pd.read_csv(r'C:\Users\rkroker\Documents\Code\Python\Students and subject teachers\Data\ENG.csv').drop(columns=columns_to_drop)
MA = pd.read_csv(r'C:\Users\rkroker\Documents\Code\Python\Students and subject teachers\Data\MA.csv').drop(columns=merge_prep)
SCI = pd.read_csv(r'C:\Users\rkroker\Documents\Code\Python\Students and subject teachers\Data\SCI.csv').drop(columns=merge_prep)
SS = pd.read_csv(r'C:\Users\rkroker\Documents\Code\Python\Students and subject teachers\Data\SS.csv').drop(columns=merge_prep)

# Rename the 'Lead Teacher | Co-Teacher' column to 'ENG Lead Teacher | Co-Teacher'
ENG = ENG.rename(columns={'Lead Teacher | Co-Teacher': 'ENG Lead Teacher | Co-Teacher'})
MA = MA.rename(columns={'Lead Teacher | Co-Teacher': 'MA Lead Teacher | Co-Teacher'})
SCI = SCI.rename(columns={'Lead Teacher | Co-Teacher': 'SCI Lead Teacher | Co-Teacher'})
SS = SS.rename(columns={'Lead Teacher | Co-Teacher': 'SS Lead Teacher | Co-Teacher'})

# Merge the dataframes on the 'LID' column
merged_df = ENG.merge(MA, on='LID', how='outer')
merged_df = merged_df.merge(SCI, on='LID', how='outer')
merged_df = merged_df.merge(SS, on='LID', how='outer')

# Reorder the columns
desired_order = [
    'Grade', 'Student', 'LID', 'ENG Lead Teacher | Co-Teacher', 
    'MA Lead Teacher | Co-Teacher', 'SCI Lead Teacher | Co-Teacher', 
    'SS Lead Teacher | Co-Teacher'
]
merged_df = merged_df[desired_order]

# Drop the LID column after merging
merged_df=merged_df.drop(columns='LID')

# Save the merged dataframe to a new CSV file
merged_df.to_csv('Students_and_Subject_Teachers.csv', index=False)

print("Merging complete. The merged file is saved as 'Students_and_Subject_Teachers.csv'.")
```
